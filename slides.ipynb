### Background
Stable isotope ratios of oxygen and hydrogen in water have long been considered powerful indicators of paleoclimate

::: {.fragment}
### Motivation
"[...] quantitative interpretation of isotope variations in terms of climate changes is hampered by a limited understanding of physical processes controlling the global isotope behavior."
:::

---

### Research approach

The autors analyzed long-term time series of:

- $^{18}\text{O}$ isotope content in monthly precipitation
- Surface air temperature

::: {.fragment}
### What are the results?

At over mid- and high-latitude regions, during the past three decades long-term changes of isotope composition in precipitation follows long-term changes of surface air temperature. The average $\delta ^{18}\text{O}$-temperature coefficient is around $0.6\text{‰} / ^\circ\text{C}$

:::



## Isotope fractionation

- **Evaporation:** preferential evaporation of lighter H₂¹⁶O → vapor enriched in ¹⁶O.
- **Condensation:** preferential incorporation of heavier H₂¹⁸O into liquid water.
- Fractionation depends on temperature.

$$
\delta^{18}\mathrm{O}
=
\left[
\frac{(^{18}\mathrm{O}/^{16}\mathrm{O})_{\mathrm{sample}}}
{(^{18}\mathrm{O}/^{16}\mathrm{O})_{\mathrm{SMOW}}}
-1
\right]\times 1000\ \text{‰}
$$

---

### Rayleigh model

**Main assumptions:**

- Isotopic equilibrium between vapor and precipitation at cloud-base temperature.
- Immediate removal of condensate from the cloud.

::: {.fragment}

**Consequences:**

- Progressive depletion of $^{18}\text{O}$ in the remaining vapor.
- Later precipitation becomes increasingly depleted in $^{18}\text{O}$.

$$
0.72 \rightarrow 0.52\ \text{‰}/^\circ\mathrm{C}
\qquad (0\text{–}20^\circ\mathrm{C})
$$

:::

### Data collection

:::: {.columns}
::: {.column width="60%"}
![IAEA/WMO global network of precipitation-isotope monitoring stations. The stations selected for long-term trend analysis are labelled.](map.png){width=100%}

<small>Source: Rozanski et al. (1992), Fig. 1.</small>
:::
::: {.column width="40%"}
**Data source**

- IAEA/WMO network *Isotopes in Precipitation* (initiated in 1961).
- Monthly precipitation $\delta^{18}\mathrm{O}$ measurements, paired with surface air temperature records.

**11 stations** were selected for the long-term trend analysis:

- **Europe (8):** Bern, Meiringen, Guttannen, Grimsel, Thonon-les-Bains, Vienna, Groningen, Krakow.
- **Other regions (3):** Ottawa, Hong Kong, Argentine Island.

<small>Selection criteria: record length and quality; continuity (precipitation in every month); representation of different climate regimes.</small>
:::
::::

---

### Time-series preprocessing

For each selected station, the authors processed the monthly $\delta^{18}\mathrm{O}$ and temperature records in the following order:

::: {.fragment}
**1. Remove the seasonal component** (12-month running average)

:::

::: {.fragment}
**2. Calculate departures from the long-term means**

$$
\Delta\delta^{18}\mathrm{O}(t),\qquad \Delta T(t)
$$

:::

::: {.fragment}
**3. Smooth the anomaly curves again** (with running average)
:::

---

### Regional composite trends

- The authors also constructed **composite trend curves** for the Swiss stations and for Europe.
- At each time step, station anomalies were combined using **arithmetic averages** of ($\Delta\delta ^{18}\mathrm{O}\) and \(\Delta T$\).
- The Swiss composite combines the four Swiss stations: Bern, Meiringen, Guttannen and Grimsel.

::: {.fragment}
**Why use composite curves?**

They help reveal regional-scale behaviour shared by several stations, rather than relying only on the record from one location.
:::

---

### Three types of $\delta ^{18}\mathrm{O}$–temperature relationships

The paper distinguishes three different comparisons:

::: {.fragment}
**(i) Spatial relationship**

Compare long-term annual mean \($\delta^{18}\mathrm{O}$\) and temperature **between different stations**.

:::

::: {.fragment}
**(ii) Seasonal relationship**

Compare short-term, seasonal changes in \($\delta^{18}\mathrm{O}$\) and temperature **at a station or across a group of stations**.

:::

::: {.fragment}
**(iii) Long-term temporal relationship**

Compare interannual changes in \($\delta^{18}\mathrm{O}$\) and temperature **at a fixed location**.

:::

## Slide with an equation

Any piecewise-continuous $2\pi$-periodic function can be approximated by a Fourier series:

$$
f(x)=\sum\limits_{n=-\infty}^{\infty} c_n e^{inx}
$$

where

$$
c_n=\frac{1}{2\pi}\int\limits_{-\pi}^{\pi} f(x)e^{-inx}\,dx
$$
